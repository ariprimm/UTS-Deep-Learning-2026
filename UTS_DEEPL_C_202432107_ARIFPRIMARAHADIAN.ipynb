```python
#202432107_ArifPrimaRahadian
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Pustaka tambahan untuk pra-pemrosesan teks dan evaluasi
import os, re, csv, copy, time, random, collections
from torch.nn.utils.rnn import pack_padded_sequence
from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                             classification_report, confusion_matrix)
from matplotlib.colors import LinearSegmentedColormap

# Seed agar hasil percobaan dapat direproduksi
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

# Gunakan GPU jika tersedia (Google Colab), jika tidak gunakan CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Versi PyTorch :', torch.__version__)
print('Device        :', device)
```

  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>

```python
#202432107_ArifPrimaRahadian
# 1. Muat dataset Anda
# File TSV dicari di folder kerja, folder dataset/, atau /content (Google Colab).
# quoting=QUOTE_NONE dan keep_default_na=False dipakai agar tanda kutip di dalam teks
# tidak merusak pembacaan baris dan kata seperti "NA" tidak diubah menjadi NaN.
def cari_file(nama):
    for folder in ['.', 'dataset', 'data', '/content', '/content/dataset']:
        path = os.path.join(folder, nama)
        if os.path.exists(path):
            return path
    raise FileNotFoundError(nama + ' tidak ditemukan. Letakkan file dataset di folder dataset/ atau folder kerja.')

def muat_tsv(nama):
    return pd.read_csv(cari_file(nama), sep='\t', quoting=csv.QUOTE_NONE, keep_default_na=False)

df_train_raw = muat_tsv('train_preprocess_ori.tsv')
df_valid_raw = muat_tsv('valid_preprocess.tsv')
df_test_raw  = muat_tsv('test_preprocess_masked_label.tsv')

# Inspeksi dataset: ukuran, kolom, distribusi label, missing value, teks kosong, duplikat
print('Nama kolom :', list(df_train_raw.columns))
label_urut = ['negative', 'neutral', 'positive']
baris = {}
for nama, d in [('train', df_train_raw), ('valid', df_valid_raw), ('test', df_test_raw)]:
    dist = d['sentiment'].value_counts()
    baris[nama] = {
        'baris': len(d),
        **{k: int(dist.get(k, 0)) for k in label_urut},
        'missing': int(d.isna().sum().sum()),
        'teks_kosong': int((d['text'].str.strip() == '').sum()),
        'duplikat_teks': int(d['text'].duplicated().sum()),
    }
print(pd.DataFrame(baris).T.to_string())
print('Label unik pada file test :', sorted(df_test_raw['sentiment'].unique()))
print('Teks train yang identik dengan teks valid :', len(set(df_train_raw['text']) & set(df_valid_raw['text'])))
print('Teks train yang identik dengan teks test  :', len(set(df_train_raw['text']) & set(df_test_raw['text'])))

# Panjang sekuens (jumlah kata per teks) pada data train
panjang_kata = df_train_raw['text'].str.split().str.len()
print('\nPanjang teks train (kata):')
print(panjang_kata.describe(percentiles=[.5, .9, .95, .99]).round(1).to_string())
negasi = ['tidak', 'bukan', 'jangan', 'belum']
print('\nJumlah teks train yang memuat kata negasi:',
      {k: int(df_train_raw['text'].str.contains(r'\b' + k + r'\b').sum()) for k in negasi})

# Visualisasi distribusi kelas dan panjang teks
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
lebar = 0.38
pos = np.arange(3)
ax[0].bar(pos - lebar/2, [baris['train'][k] for k in label_urut], lebar, color='#114F6A', label='Train')
ax[0].bar(pos + lebar/2, [baris['valid'][k] for k in label_urut], lebar, color='#1FA4B1', label='Valid')
for i, k in enumerate(label_urut):
    ax[0].text(i - lebar/2, baris['train'][k], str(baris['train'][k]), ha='center', va='bottom', fontsize=9)
    ax[0].text(i + lebar/2, baris['valid'][k], str(baris['valid'][k]), ha='center', va='bottom', fontsize=9)
ax[0].set_xticks(pos)
ax[0].set_xticklabels(label_urut)
ax[0].set_title('Distribusi Kelas Sentimen')
ax[0].set_ylabel('Jumlah teks')
ax[0].legend()
ax[1].hist(panjang_kata, bins=40, color='#114F6A', alpha=0.9)
ax[1].axvline(panjang_kata.quantile(0.99), color='#1FA4B1', linestyle='--', linewidth=2, label='Persentil ke-99')
ax[1].set_title('Distribusi Panjang Teks (Data Train)')
ax[1].set_xlabel('Jumlah kata')
ax[1].set_ylabel('Jumlah teks')
ax[1].legend()
for a in ax:
    a.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
#202432107_ArifPrimaRahadian
# 2. Lakukan pembersihan data & pra-pemrosesan (NLP): case folding, tokenisasi,
#    vocabulary mapping, padding, dan label encoding

# a. Buang teks duplikat pada train dan teks train yang sama dengan data uji (mencegah kebocoran data)
df_pool = df_train_raw.drop_duplicates(subset='text')
n_dup = len(df_train_raw) - len(df_pool)
df_pool = df_pool[~df_pool['text'].isin(set(df_valid_raw['text']))].reset_index(drop=True)
n_bocor = len(df_train_raw) - n_dup - len(df_pool)
print('Train asli: {} | duplikat dibuang: {} | beririsan dengan data uji dibuang: {} | tersisa: {}'.format(
      len(df_train_raw), n_dup, n_bocor, len(df_pool)))

# b. Tokenisasi dengan case folding. Kata negasi (tidak, bukan, dst.) sengaja dipertahankan
#    karena mengubah makna sentimen. Tanda seru dan tanya dipertahankan sebagai token.
def tokenisasi(teks):
    teks = teks.lower()
    return re.findall(r"[a-z0-9]+(?:[-'][a-z0-9]+)*|[!?]", teks)

# c. Bagi data: 90% latih dan 10% validasi (stratified). Data uji berlabel = file valid asli.
df_tr, df_val = train_test_split(df_pool, test_size=0.1, stratify=df_pool['sentiment'], random_state=SEED)
df_tr = df_tr.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_uji = df_valid_raw.copy()

# d. Bangun vocabulary HANYA dari data latih (token jarang dengan frekuensi < MIN_FREQ menjadi <unk>)
PAD, UNK = '<pad>', '<unk>'
MIN_FREQ = 2
hitung = collections.Counter(tok for teks in df_tr['text'] for tok in tokenisasi(teks))
vocab = {PAD: 0, UNK: 1}
for kata, frek in hitung.most_common():
    if frek >= MIN_FREQ:
        vocab[kata] = len(vocab)
print('Ukuran vocabulary (termasuk <pad> dan <unk>):', len(vocab))

# e. Label encoding
label2id = {'negative': 0, 'neutral': 1, 'positive': 2}
id2label = {v: k for k, v in label2id.items()}
nama_kelas = [id2label[i] for i in range(3)]

# f. Token-to-index encoding dan padding ke panjang tetap (post-padding dengan indeks 0)
MAX_LEN = 80
def encode_teks(daftar_teks):
    urutan = [[vocab.get(t, vocab[UNK]) for t in tokenisasi(x)][:MAX_LEN] for x in daftar_teks]
    urutan = [u if len(u) > 0 else [vocab[UNK]] for u in urutan]
    panjang = torch.tensor([len(u) for u in urutan], dtype=torch.long)
    X = torch.zeros(len(urutan), MAX_LEN, dtype=torch.long)
    for i, u in enumerate(urutan):
        X[i, :len(u)] = torch.tensor(u, dtype=torch.long)
    return X, panjang

def tingkat_oov(daftar_teks):
    token = [t for x in daftar_teks for t in tokenisasi(x)]
    return sum(t not in vocab for t in token) / len(token)

# 3. Konversi data menjadi PyTorch Tensor (X_train, y_train, X_test, y_test)
X_train, L_train = encode_teks(df_tr['text'])
X_val,   L_val   = encode_teks(df_val['text'])
X_test,  L_test  = encode_teks(df_uji['text'])
X_unl,   L_unl   = encode_teks(df_test_raw['text'])      # data tanpa label (label asli disamarkan)
y_train = torch.tensor(df_tr['sentiment'].map(label2id).values, dtype=torch.long)
y_val   = torch.tensor(df_val['sentiment'].map(label2id).values, dtype=torch.long)
y_test  = torch.tensor(df_uji['sentiment'].map(label2id).values, dtype=torch.long)

batch_size = 64
g = torch.Generator()
g.manual_seed(SEED)
train_loader = DataLoader(TensorDataset(X_train, L_train, y_train), batch_size=batch_size, shuffle=True, generator=g)

n_potong = sum(len(tokenisasi(x)) > MAX_LEN for x in df_tr['text'])
print('\nBentuk tensor -> X_train: {}, X_val: {}, X_test: {}, X_unl: {}'.format(
      tuple(X_train.shape), tuple(X_val.shape), tuple(X_test.shape), tuple(X_unl.shape)))
print('Teks train yang terpotong oleh MAX_LEN={}: {} ({:.2f}%)'.format(MAX_LEN, n_potong, 100 * n_potong / len(df_tr)))
print('Tingkat OOV -> train: {:.2%}, valid: {:.2%}, test: {:.2%}'.format(
      tingkat_oov(df_tr['text']), tingkat_oov(df_val['text']), tingkat_oov(df_uji['text'])))
print('Distribusi label train :', {id2label[k]: int(v) for k, v in enumerate(torch.bincount(y_train))})
print('Distribusi label valid :', {id2label[k]: int(v) for k, v in enumerate(torch.bincount(y_val))})
print('Distribusi label test  :', {id2label[k]: int(v) for k, v in enumerate(torch.bincount(y_test))})
print('\nContoh tokenisasi :', tokenisasi(df_tr['text'][0])[:12])
print('Contoh indeks     :', X_train[0][:12].tolist())
```

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432107_ArifPrimaRahadian
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_size, num_classes, dropout=0.5, pad_idx=0):
        super(SequenceModel, self).__init__()
        # Embedding -> LSTM -> Dropout -> Fully Connected
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=pad_idx)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x, lengths):
        # x: (batch, MAX_LEN) berisi indeks kata, lengths: panjang asli tiap teks tanpa padding
        embedded = self.embedding(x)
        packed = pack_padded_sequence(embedded, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, (h_n, c_n) = self.lstm(packed)       # h_n = hidden state pada kata terakhir yang asli
        ringkasan = self.dropout(h_n[-1])       # (batch, hidden_size)
        return self.fc(ringkasan)               # logit untuk 3 kelas sentimen
```

  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432107_ArifPrimaRahadian
# 1. Inisiasi hyperparameter (epochs, learning_rate, hidden_size, dll.)
# Nilai contoh pada template (epochs=50, learning_rate=0.01) disesuaikan dengan kasus ini:
# learning rate 0.001 lebih aman untuk Adam pada LSTM, dan jumlah epoch dibatasi
# karena pelatihan dihentikan lebih awal (early stopping) bila loss validasi tidak membaik.
epochs = 15
learning_rate = 0.001
embed_dim = 128
hidden_size = 128
dropout_rate = 0.5
patience = 3

# 2. Inisiasi objek model, fungsi Loss (CrossEntropyLoss), dan Optimizer (Adam)
model = SequenceModel(len(vocab), embed_dim, hidden_size, 3, dropout_rate).to(device)

# Bobot kelas (kebalikan frekuensi) agar kelas neutral yang jarang tidak diabaikan model
jumlah_kelas = torch.bincount(y_train).float()
bobot_kelas = len(y_train) / (3 * jumlah_kelas)
print('Bobot kelas', nama_kelas, ':', [round(b, 3) for b in bobot_kelas.tolist()])
criterion = nn.CrossEntropyLoss(weight=bobot_kelas.to(device))
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
print(model)
print('Jumlah parameter:', sum(p.numel() for p in model.parameters()))

def predict_logits(model, X, L, batch=256):
    model.eval()
    hasil = []
    with torch.no_grad():
        for i in range(0, len(X), batch):
            hasil.append(model(X[i:i+batch].to(device), L[i:i+batch]).cpu())
    return torch.cat(hasil)

# 3. Simpan riwayat nilai loss dan accuracy pada list untuk divisualisasikan
train_losses, val_losses = [], []
train_accs, val_accs = [], []
best_val_loss, best_epoch, best_state, tanpa_perbaikan = float('inf'), 0, None, 0

# 4. Jalankan proses perulangan training (Training Loop)
mulai = time.time()
for epoch in range(1, epochs + 1):
    model.train()
    total_loss, benar, total = 0.0, 0, 0
    for xb, lb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        logit = model(xb, lb)                    # forward propagation
        loss = criterion(logit, yb)              # perhitungan loss
        loss.backward()                          # backpropagation
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()                         # pembaruan bobot
        total_loss += loss.item() * len(yb)
        benar += (logit.argmax(1) == yb).sum().item()
        total += len(yb)
    train_losses.append(total_loss / total)
    train_accs.append(benar / total)

    # Validasi pada data validasi
    logit_val = predict_logits(model, X_val, L_val)
    val_loss = criterion(logit_val.to(device), y_val.to(device)).item()
    val_losses.append(val_loss)
    val_accs.append((logit_val.argmax(1) == y_val).float().mean().item())

    print('Epoch {:2d}/{} | Train Loss {:.4f} Acc {:.4f} | Val Loss {:.4f} Acc {:.4f}'.format(
          epoch, epochs, train_losses[-1], train_accs[-1], val_losses[-1], val_accs[-1]))

    # Simpan bobot terbaik dan hentikan lebih awal bila tidak ada perbaikan
    if val_loss < best_val_loss:
        best_val_loss, best_epoch, tanpa_perbaikan = val_loss, epoch, 0
        best_state = copy.deepcopy(model.state_dict())
    else:
        tanpa_perbaikan += 1
        if tanpa_perbaikan >= patience:
            print('Early stopping pada epoch', epoch)
            break

model.load_state_dict(best_state)
print('\nBobot terbaik dari epoch {} (Val Loss {:.4f}). Waktu pelatihan: {:.1f} detik'.format(
      best_epoch, best_val_loss, time.time() - mulai))
```
